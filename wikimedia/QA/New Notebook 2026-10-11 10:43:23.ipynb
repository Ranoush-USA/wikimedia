```python
%sql
select 'silver' as source_table,
       min(try_to_date(timestamp)) as first_occurrence,
       max(try_to_date(timestamp)) as last_occurrence,
       count(*) as total_events
from workspace.silver.wikimedia_silver

union all

select 'bronze' as source_table,
       min(try_to_date(timestamp, 'yyyyMMdd')) as first_occurrence,
       max(try_to_date(timestamp, 'yyyyMMdd')) as last_occurrence,
       count(*) as total_events
from workspace.bronze.wikimedia_bronze;
```
